# Appendix C: Frameworks, Standards, and Reference Documents

This appendix summarises the key frameworks, standards, and reference documents cited throughout
the book, with pointers to official sources.

---

## Risk and Control Frameworks

### NIST Cybersecurity Framework (CSF) 2.0

**Publisher**: National Institute of Standards and Technology  
**URL**: https://www.nist.gov/cyberframework  
**Scope**: Voluntary framework for managing cybersecurity risk across all sectors.  
**Key feature**: Six functions: Govern, Identify, Protect, Detect, Respond, Recover.  
**Used in**: Chapter 1

### NIST Risk Management Framework (RMF)

**Publisher**: NIST  
**URL**: https://csrc.nist.gov/projects/risk-management  
**Scope**: Structured six-step risk management process; mandatory for US federal systems.  
**Used in**: Chapter 5

### NIST SP 800-30: Guide for Conducting Risk Assessments

**Publisher**: NIST  
**URL**: https://csrc.nist.gov/publications/detail/sp/800-30/rev-1/final  
**Used in**: Chapter 5

### NIST SP 800-53: Security and Privacy Controls

**Publisher**: NIST  
**URL**: https://csrc.nist.gov/publications/detail/sp/800-53/rev-5/final  
**Scope**: Comprehensive catalog of security and privacy controls for federal systems.  
**Used in**: Chapter 5

### NIST SP 800-61: Computer Security Incident Handling Guide

**Publisher**: NIST  
**URL**: https://csrc.nist.gov/publications/detail/sp/800-61/rev-2/final  
**Used in**: Chapter 14

### NIST SP 800-82: Guide to ICS Security

**Publisher**: NIST  
**URL**: https://csrc.nist.gov/publications/detail/sp/800-82/rev-3/final  
**Used in**: Chapter 20

## Security Standards

### ISO/IEC 27001: Information Security Management Systems

**Publisher**: ISO/IEC  
**URL**: https://www.iso.org/standard/27001  
**Scope**: International standard for an ISMS; provides a systematic approach to managing information security risk.  
**Used in**: Chapter 5

### IEC 62443: Industrial Automation and Control System Security

**Publisher**: IEC  
**Scope**: Comprehensive standard for OT/ICS security covering operators, integrators, and component manufacturers.  
**Used in**: Chapter 20

### PCI DSS: Payment Card Industry Data Security Standard

**Publisher**: PCI Security Standards Council  
**URL**: https://www.pcisecuritystandards.org  
**Scope**: Security requirements for organisations that handle payment card data.  
**Used in**: Chapter 18

### MITRE ATT&CK

**Publisher**: MITRE Corporation  
**URL**: https://attack.mitre.org  
**Scope**: Knowledge base of adversary tactics, techniques, and procedures (TTPs).  
**Used in**: Chapters 9, 14, 15

### OWASP Top 10

**Publisher**: Open Web Application Security Project  
**URL**: https://owasp.org/Top10  
**Scope**: Ten most critical web application security risk categories.  
**Used in**: Chapter 10

## Vulnerability Databases

### National Vulnerability Database (NVD)

**URL**: https://nvd.nist.gov  
**Scope**: US government repository of vulnerability data based on CVE identifiers.

### CVE (Common Vulnerabilities and Exposures)

**Publisher**: MITRE Corporation  
**URL**: https://cve.mitre.org  
**Scope**: Dictionary of publicly known vulnerabilities.

### CWE (Common Weakness Enumeration)

**Publisher**: MITRE Corporation  
**URL**: https://cwe.mitre.org  
**Scope**: Categorised list of software and hardware weakness types.

### EXPLOIT-DB

**URL**: https://www.exploit-db.com  
**Scope**: Archive of public exploits and vulnerable software maintained by Offensive Security.
