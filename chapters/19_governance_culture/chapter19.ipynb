## Key Terms

- **Policy**: a high-level mandatory statement of management intent, the "what" of security.
- **Standard**: a mandatory, specific rule that implements a policy, the "how much."
- **Guideline**: a recommended, non-mandatory practice.
- **Procedure**: a step-by-step instruction, the "how."
- **RACI**: Responsible, Accountable, Consulted, Informed.
- **KGI**: Key Goal Indicator, measures outcome achievement.
- **KPI**: Key Performance Indicator, measures process performance.
- **GRC**: Governance, Risk, and Compliance.
- **COBIT**: Control Objectives for Information and Related Technologies, a governance framework from ISACA.

## 19.1 Governance Versus Management

Governance sets direction and assigns accountability; management executes within that direction.
Governance answers "what are we trying to achieve and who is answerable," while management answers "how do
we get there day to day." The distinction matters because security failures are frequently governance
failures: the control existed, but no one owned it, funded it, or was accountable when it lapsed.

## 19.2 The Policy Document Hierarchy

Security direction flows downward through a hierarchy of documents, each more specific than the one above:

- **Policies** are high-level and mandatory. They state intent, for example "all sensitive data must be
  encrypted at rest." They change rarely and are approved by senior leadership.
- **Standards** are mandatory and specific. They quantify a policy, for example "encryption at rest must use
  AES-256." They change as technology changes.
- **Guidelines** are recommended but not mandatory. They offer advice for situations the standards do not
  fully cover.
- **Procedures** are step-by-step instructions that tell a specific person how to perform a task, for
  example the exact commands to rotate a key.

A common failure is writing procedures without the policy above them, which leaves staff following steps
nobody can connect to a business rationale, or writing policies with no procedures below them, which leaves
intent that no one knows how to carry out.

## 19.3 Assigning Responsibility with RACI

A RACI chart maps tasks to roles so that accountability is never ambiguous. For each task exactly one role
is **Accountable** (the single owner who answers for the outcome), one or more are **Responsible** (those
who do the work), others may be **Consulted** (whose input is sought), and others **Informed** (who are kept
up to date). The discipline of allowing only one accountable party per task is what prevents the
"everyone assumed someone else was handling it" class of incident.

## 19.4 Measuring Governance: KGIs and KPIs

Governance frameworks such as COBIT distinguish two kinds of metric. A **Key Goal Indicator** measures
whether an objective was achieved, the outcome, for example "percentage of systems patched within the
service-level window." A **Key Performance Indicator** measures how well a process is executing, the
leading signal, for example "mean time to apply a critical patch." KGIs tell leadership whether the program
is working; KPIs tell operators whether the process that produces that result is healthy.

## 19.5 Security Culture

A security culture exists when information security is a shared value embedded in an organization's mission
and daily behavior rather than a compliance burden enforced by the IT department. In a strong culture,
employees understand why security matters, feel personally responsible, and actively report threats. The
SANS Security Awareness Maturity Model describes five levels:

1. **Non-Existent**: no formal program; incidents are the only signal.
2. **Compliance-Focused**: annual training to satisfy a checkbox; completion is high but behavior does not
   change and phishing click rates stay high.
3. **Promoting Awareness and Behavior Change**: regular, engaging training and phishing simulations with
   metrics tracked.
4. **Long-Term Sustainment and Culture Change**: security is reinforced continuously and woven into
   processes.
5. **Metrics Framework**: the program is measured against business outcomes and improved with data.

The model is useful because it separates the appearance of awareness (high training-completion rates) from
its substance (changed behavior under real pressure).

## 19.6 Why This Matters

Most controls ultimately depend on people choosing to use them. A perfectly engineered system can be undone
by one approval given under social-engineering pressure, as the human-factors chapter showed. Governance and
culture are the mechanisms that make secure behavior the path of least resistance rather than an act of
individual heroism.

## 19.7 News in Focus

Several large breaches in the last decade were later characterized by investigators as governance failures
rather than novel technical attacks: a known vulnerability went unpatched because no team clearly owned the
asset, or an alert fired but no defined procedure routed it to anyone with authority to act. These cases are
why frameworks emphasize ownership and accountability as heavily as technical controls.

## 19.8 Worked Example: Scoring Awareness Maturity

The function below turns a few observable signals into an approximate maturity level, illustrating how KPIs
feed a KGI-style assessment.


````python
def maturity_level(has_program, training_completion, behavior_tracked,
                   click_rate, business_metrics):
    # Signals are deliberately simple proxies for the SANS five-level model
    if not has_program:
        return 1, "Non-Existent"
    if training_completion >= 0.9 and not behavior_tracked:
        return 2, "Compliance-Focused"
    if behavior_tracked and click_rate > 0.05 and not business_metrics:
        return 3, "Promoting Awareness and Behavior Change"
    if behavior_tracked and click_rate <= 0.05 and not business_metrics:
        return 4, "Long-Term Sustainment and Culture Change"
    if business_metrics and click_rate <= 0.05:
        return 5, "Metrics Framework"
    return 3, "Promoting Awareness and Behavior Change"

orgs = [
    dict(name="Acme",   has_program=True,  training_completion=0.95, behavior_tracked=False, click_rate=0.18, business_metrics=False),
    dict(name="Globex", has_program=True,  training_completion=0.92, behavior_tracked=True,  click_rate=0.09, business_metrics=False),
    dict(name="Initech",has_program=True,  training_completion=0.98, behavior_tracked=True,  click_rate=0.03, business_metrics=True),
]

for o in orgs:
    name = o.pop("name")
    level, label = maturity_level(**o)
    print(f"{name:8} -> Level {level}: {label}")
````
