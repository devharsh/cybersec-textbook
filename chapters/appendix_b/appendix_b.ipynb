# Appendix B: Common Ports and Protocols

A working knowledge of common ports speeds up scanning, enumeration, and troubleshooting. The table
below lists widely used services. Always verify against the authoritative IANA registry for production
work.

| Port | Protocol | Service | Notes |
|---|---|---|---|
| 20, 21 | TCP | FTP | Unencrypted; prefer SFTP or FTPS |
| 22 | TCP | SSH | Secure remote administration |
| 23 | TCP | Telnet | Unencrypted; deprecated |
| 25 | TCP | SMTP | Mail transfer |
| 53 | TCP/UDP | DNS | Name resolution |
| 67, 68 | UDP | DHCP | Address assignment |
| 80 | TCP | HTTP | Unencrypted web |
| 110 | TCP | POP3 | Mail retrieval |
| 123 | UDP | NTP | Time synchronization |
| 143 | TCP | IMAP | Mail retrieval |
| 161, 162 | UDP | SNMP | Network management |
| 389 | TCP/UDP | LDAP | Directory services |
| 443 | TCP | HTTPS | Encrypted web (TLS) |
| 445 | TCP | SMB | Windows file sharing |
| 514 | UDP | Syslog | Log transport |
| 587 | TCP | SMTP (submission) | Authenticated mail submission |
| 636 | TCP | LDAPS | LDAP over TLS |
| 993 | TCP | IMAPS | IMAP over TLS |
| 995 | TCP | POP3S | POP3 over TLS |
| 3306 | TCP | MySQL | Database |
| 3389 | TCP | RDP | Remote Desktop |
| 5432 | TCP | PostgreSQL | Database |
| 8080 | TCP | HTTP alt | Common proxy or app server |

A useful mental shortcut: many secure variants run on a nearby port and add TLS, such as HTTP on 80
versus HTTPS on 443, or LDAP on 389 versus LDAPS on 636.
