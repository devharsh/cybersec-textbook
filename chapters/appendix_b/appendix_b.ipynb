# Appendix B: Glossary

This glossary defines key terms used throughout the book. Terms are cross-referenced to the
chapter where they are introduced in depth.

---

**Access control list (ACL)**: An ordered set of permit and deny rules applied to network traffic or file-system permissions. (Ch. 11)

**Advanced Persistent Threat (APT)**: A sophisticated, long-duration threat actor, typically state-sponsored, with significant resources and patience. (Ch. 1)

**Air gap**: Physical separation between two networks with no network connection. (Ch. 20)

**Annualized Loss Expectancy (ALE)**: Expected monetary loss from a risk per year; ALE = SLE × ARO. (Ch. 5)

**Anomaly detection**: Detecting deviations from a baseline of normal behavior. (Ch. 12)

**Attack surface**: The total set of exploitable entry points in a system. (Ch. 1)

**Authentication**: Verifying the identity of a user, system, or process. (Ch. 2)

**Availability**: The property that systems and data are accessible when needed. (Ch. 1)

**Baiting**: Leaving a malicious device (USB) for a victim to use. (Ch. 4)

**Bitcoin**: A decentralized digital currency frequently used for ransomware payments. (Ch. 15)

**Blue team**: Defenders; the team responsible for protecting systems. (Ch. 6)

**Botnet**: A network of compromised machines under attacker control. (Ch. 15)

**Buffer overflow**: Writing past the end of an allocated memory buffer. (Ch. 9)

**Business Email Compromise (BEC)**: Impersonating an executive to authorize fraudulent transactions. (Ch. 4)

**CISO**: Chief Information Security Officer. (Ch. 19)

**CIA triad**: Confidentiality, Integrity, Availability; the three core security properties. (Ch. 1)

**Certificate Authority (CA)**: A trusted entity that issues digital certificates. (Ch. 2)

**Chain of custody**: Documentation of who handled evidence and when. (Ch. 13)

**CVSS**: Common Vulnerability Scoring System; a 0-10 severity scale for vulnerabilities. (Ch. 6)

**CVE**: Common Vulnerabilities and Exposures; a unique identifier for a known vulnerability. (Ch. 8)

**Confidential virtual machine**: A virtual machine whose memory and processor state hardware protects from the hypervisor and the cloud operator, such as an Intel TDX trust domain or an AMD SEV-SNP guest. (Ch. 1)

**Confidentiality**: Preventing unauthorized access to information. (Ch. 1)

**Cryptography**: Mathematical techniques for secure communication and data protection. (Ch. 2)

**Data minimization**: Collecting only the personal data necessary for the stated purpose. (Ch. 18)

**Defense in depth**: Using multiple overlapping security layers. (Ch. 1)

**Digital signature**: A cryptographic proof of origin and integrity using asymmetric keys. (Ch. 2)

**DMZ**: Demilitarized Zone; a network segment between the internet and internal network. (Ch. 11)

**DDoS**: Distributed Denial of Service; overwhelming a system with traffic from many sources. (Ch. 11)

**Encryption**: Transforming data with a key so only key holders can reverse it. (Ch. 2)

**Exploit**: Code or technique that takes advantage of a vulnerability. (Ch. 1, 9)

**False positive**: An alert fired on benign activity. (Ch. 12)

**False negative**: A real attack that did not trigger an alert. (Ch. 12)

**Firewall**: A device or software that enforces access control between networks. (Ch. 11)

**Forensic image**: A bit-for-bit copy of a storage device. (Ch. 13)

**GDPR**: General Data Protection Regulation; EU law governing personal data. (Ch. 18)

**Hash function**: A one-way function producing a fixed-length digest. (Ch. 2)

**Heavy tail**: A distribution tail that thins out more slowly than any exponential curve, as with power laws and lognormals, so extreme values are far more common than a normal curve predicts; when the tail is heavy enough, averages are unstable and a few cases dominate the total. (Ch. 17)

**HIPAA**: Health Insurance Portability and Accountability Act; US healthcare privacy law. (Ch. 18)

**HMAC**: Hash-based Message Authentication Code. (Ch. 2)

**ICS**: Industrial Control System. (Ch. 20)

**IDS/IPS**: Intrusion Detection System / Intrusion Prevention System. (Ch. 12)

**Incident**: An event that jeopardises confidentiality, integrity, or availability. (Ch. 14)

**Integrity**: Assurance that data has not been altered without authorization. (Ch. 1)

**IOC**: Indicator of Compromise. (Ch. 14, 15)

**Lateral movement**: Moving from an initial foothold to other network systems. (Ch. 9)

**Malware**: Malicious software. (Ch. 15)

**Modulo bias**: Non-uniformity introduced when random values are reduced to a smaller range with the remainder operation and the range does not divide their count evenly; avoided by rejection sampling. (Ch. 17)

**MTTD**: Mean Time to Detect. (Ch. 12, 14)

**NIST CSF**: NIST Cybersecurity Framework. (Ch. 1)

**Normal (Gaussian) distribution**: The symmetric bell curve fixed by a mean and a standard deviation; the standard normal distribution has mean 0 and standard deviation 1. (Ch. 17)

**OSINT**: Open-Source Intelligence. (Ch. 7)

**OWASP Top 10**: The ten most critical web application security risk categories. (Ch. 10)

**Pareto distribution**: A heavy-tailed distribution whose upper tail falls off as a power of the value, a common model for wealth and many loss sizes; with a tail index near 1.16 it produces the 80/20 rule. (Ch. 17)

**Pass-the-hash**: Authenticating with a captured NTLM hash without cracking it. (Ch. 9)

**PCI DSS**: Payment Card Industry Data Security Standard. (Ch. 18)

**Penetration test**: An authorized simulated attack to identify exploitable vulnerabilities. (Ch. 6)

**Phishing**: Mass email impersonating a trusted entity to steal credentials. (Ch. 4)

**PKI**: Public Key Infrastructure. (Ch. 2)

**PLC**: Programmable Logic Controller. (Ch. 20)

**Poisson distribution**: The distribution of the number of independent events in a fixed window at a steady average rate; its variance equals its mean. (Ch. 5, 17)

**Post-quantum cryptography**: Cryptographic algorithms resistant to quantum computers. (Ch. 17)

**Privilege escalation**: Gaining higher access than initially obtained. (Ch. 9)

**Ransomware**: Malware that encrypts files and demands payment. (Ch. 15)

**Red team**: Attackers; the team simulating adversary behavior. (Ch. 6)

**Risk**: The probability that a threat exploits a vulnerability, weighted by impact. (Ch. 5)

**Risk compensation**: The tendency to take more risk when protected, which can erode the benefit of a control. (Ch. 5)

**ROSI**: Return on Security Investment. (Ch. 5)

**RPO**: Recovery Point Objective; maximum tolerable data loss window. (Ch. 5)

**RTO**: Recovery Time Objective; maximum tolerable downtime. (Ch. 5)

**SCADA**: Supervisory Control and Data Acquisition. (Ch. 20)

**SIEM**: Security Information and Event Management. (Ch. 12)

**Social engineering**: Manipulating people to divulge information or take actions. (Ch. 4)

**SQL injection**: Injecting SQL code via user input to manipulate a database. (Ch. 10)

**Supply chain attack**: Compromising a target by attacking a trusted vendor or dependency. (Ch. 17)

**System Management Mode (SMM)**: An x86 processor mode, entered only through a system management interrupt, in which firmware runs from protected memory that the operating system cannot see; informally called ring -2. (Ch. 1)

**Threat actor**: A person or group responsible for an attack. (Ch. 1)

**TLS**: Transport Layer Security; protocol securing HTTPS and other communications. (Ch. 2, 11)

**Trusted execution environment (TEE)**: An area that hardware protects even from more privileged software, such as an SGX enclave, the Arm TrustZone secure world, or a confidential virtual machine. (Ch. 1)

**Trusted platform module (TPM)**: A root of trust that generates and protects keys and records boot measurements; it may be a discrete chip, integrated circuitry, or firmware. (Ch. 1)

**Uniform distribution**: A distribution in which every value in a range is equally likely, such as a fair die; the target for cryptographic keys, tokens and other random secrets. (Ch. 17)

**Vulnerability**: A weakness that can be exploited. (Ch. 1)

**YARA**: A pattern-matching language for malware identification. (Ch. 15)

**Z-score**: The number of standard deviations a value lies from the mean. (Ch. 17)

**Zero trust**: Security model requiring verification for every request regardless of network location. (Ch. 11)

**Zero-day**: A vulnerability unknown to the vendor; no patch exists. (Ch. 17)

## Adversary and Threat-Model Terminology

These terms specify *what the attacker is allowed to do*, and they recur throughout the cryptography
(Chapter 2), risk (Chapter 5), and privacy-preserving computation (Chapter 17) material. Appendix J develops
them in full, with the defining papers, worked numbers, and exercises.

**Adversary model**: A precise statement of an attacker's behavior, corruption powers, and winning condition,
without which a security claim is meaningless. (App. J)

**Semi-honest (honest-but-curious, passive) adversary**: A party that follows the protocol exactly but tries
to infer extra information from everything it legitimately sees (its input, random coins, and received
messages). (App. J, Ch. 17)

**Honest-but-lazy (lazy) adversary**: A party that behaves honestly but aborts partway through, or skips
costly work, without colluding with corrupt parties. Its applied form in collaborative learning is the
free-rider. (App. J)

**Free-rider**: A federated-learning client that submits fabricated updates without actually training, to
obtain the global model without contributing. (App. J, Ch. 17)

**Fail-stop / omission corruption**: An adversary power that crashes a party outright, or selectively blocks
its messages, without reading its state. (App. J)

**Honest-looking adversary**: One that deviates only in ways indistinguishable from honest execution.
(App. J)

**Semi-malicious adversary**: A party that must send protocol-conformant messages but may choose its input
and random coins freely; the standard model for multi-key and threshold homomorphic encryption. (App. J,
Ch. 17)

**Covert adversary**: One that may cheat arbitrarily but is caught with at least a fixed probability
(the deterrence factor), so cheating is irrational for a reputation-sensitive party. (App. J)

**Deterrence factor ($\epsilon$)**: The guaranteed probability of detecting cheating in a covert-secure
protocol. (App. J)

**Rational adversary**: A utility-maximizing party that cheats only when the expected payoff exceeds the
expected penalty. (App. J)

**Malicious (active, Byzantine) adversary**: A party wholly controlled by the attacker, free to deviate
arbitrarily, forge inputs and keys, or abort. (App. J, Ch. 2)

**Mixed adversary**: One corrupting some parties actively, others passively, and others in a fail-stop
manner, simultaneously. (App. J)

**Friends-and-foes (FaF) security**: A requirement that honest parties' inputs stay hidden even from other,
curious honest parties who receive leaked adversary views. (App. J)

**Static / adaptive / mobile adversary**: One that fixes its corruption set before the protocol, chooses it
during execution, or moves between parties over time (countered by proactive share refresh). (App. J)

**Rushing adversary**: One that waits to see the honest parties' round messages before sending its own.
(App. J)

**IND-CPA / IND-CCA1 / IND-CCA2**: Indistinguishability under chosen-plaintext attack, under a
non-adaptive ("lunchtime") chosen-ciphertext attack, and under an adaptive chosen-ciphertext attack, in
increasing order of strength. (Ch. 2, App. J)

**IND-CPA$^{\mathbf{D}}$** (also written IND-CPA-D): Indistinguishability under chosen-plaintext attack *with decryption
oracles* on honestly evaluated ciphertexts; the correct passive notion for homomorphic encryption, since plain
IND-CPA proved inadequate for approximate schemes such as CKKS. (Ch. 2, Ch. 17, App. J)

**Non-malleability (NM)**: The requirement that an attacker cannot transform a ciphertext into one whose
plaintext is meaningfully related; unachievable for homomorphic encryption by design. (Ch. 2, App. J)

**Smudging noise**: Extra noise added after (partial) decryption in threshold or homomorphic schemes to
prevent the decryption result from leaking the secret key. (App. J, Ch. 17)

**One-sided simulation**: A relaxed malicious-security notion giving full simulation for the party that
receives the output and privacy for the other; used in efficient two-party protocols. (App. J, Ch. 17)

**Collusion resistance**: The requirement that users pooling their decryption keys gain no capability beyond
what each holds individually; central to identity-, attribute-, and functional encryption. (App. J)

### Executable formats and file structure

These entries cover the vocabulary of executable files: the Windows and Unix formats, how addresses inside a loaded image are calculated, the tables and entry points that reveal packing, and the header-controlled protections and per-thread data an analyst meets when opening a binary.

**Portable Executable (PE)**: The Windows executable and object file format. PE32 and PE32+ are its
32-bit and 64-bit forms, distinguished by the optional header magic value. (Ch. 15)

**Common Object File Format (COFF)**: The older object format the PE format was built on. The file
header inside a PE file is still a COFF header. (Ch. 15)

**Executable and Linkable Format (ELF)**: The Unix and Linux executable format. (Ch. 15)

**Relative Virtual Address (RVA)**: An offset from the image base once a file is loaded. A virtual
address is the image base plus the RVA; a file offset is the RVA minus the section's virtual address
plus that section's pointer to raw data. (Ch. 15)

**Import Address Table (IAT)**: The array the loader fills with the resolved addresses of imported
functions. A nearly empty one is a packing indicator. The Import Name Table (INT) is the parallel
array naming those functions, and it is unrelated to the x86 INT instruction. (Ch. 15)

**Original entry point (OEP)**: In a packed program, the address where the unpacking stub finally
transfers control to the real code. (Ch. 15)

**Address Space Layout Randomization (ASLR)** and **Data Execution Prevention (DEP)**: Load-time
address randomization and non-executable data pages. Both are opted into through PE header flags.
(Ch. 15)

**Thread Local Storage (TLS)**: Per-thread global data. A PE file's TLS callbacks run before the
entry point, which makes them an anti-debugging vehicle. Not Transport Layer Security. (Ch. 15)

### Processor and architecture

The terms below concern the processor beneath the code: the calling conventions that differ between Windows and Linux, the instructions malware uses to detect a hypervisor or debugger, and the hardware support that lets guest code run on the real processor.

**Application Binary Interface (ABI)**: The conventions above the instruction set: which register
carries which argument, who cleans up the stack, how objects are laid out. It differs between Windows
and Linux, which is why calling conventions must be named rather than assumed. (Ch. 15)

**CPUID and RDTSC**: The processor identification instruction and the read time-stamp counter
instruction. Both are used by malware to detect a hypervisor or a debugger. (Ch. 15)

**Intel VT-x and AMD-V**: Hardware virtualization support, which lets guest instructions execute on
the real processor rather than being emulated. (Ch. 15)

### Windows internals

These Windows structures and interfaces come up repeatedly in malware analysis, from the per-process and per-thread blocks behind anti-debugging checks to queued procedure calls, management interfaces, and C++ type metadata that can survive in a stripped binary.

**Process Environment Block (PEB)** and **Thread Environment Block (TEB)**: Per-process and
per-thread structures in user memory. The PEB's BeingDebugged byte is the simplest anti-debugging
check there is, and shellcode reaches the PEB through the TEB with no imports at all. (Ch. 15)

**Asynchronous Procedure Call (APC)**: A function queued to run on a specific thread, used as a
covert launching technique. (Ch. 15)

**Windows Management Instrumentation (WMI)** and **Component Object Model (COM)**: Management query
and object interfaces that let malware reach system functionality without an obvious import. (Ch. 15)

**Run-Time Type Information (RTTI)**: C++ metadata that survives compilation and often recovers class
names and inheritance from a stripped binary. (Ch. 15)

### Tools and projects

The entries below expand the names of common reverse engineering tools and projects, from the FLARE team's string solver and the UPX packer to debugger add-ons, binary libraries, and service simulators, and they note which names are product names rather than acronyms.

**FireEye Labs Advanced Reverse Engineering (FLARE)**: The team, now inside Google Cloud through
Mandiant, that publishes FLARE-VM, FLOSS and capa. (Ch. 15)

**FLARE Obfuscated String Solver (FLOSS)**: Formerly the FireEye Labs Obfuscated String Solver.
Recovers strings that ordinary extraction misses because they are constructed or decoded at run time.
(Ch. 15)

**Ultimate Packer for eXecutables (UPX)**: The open-source packer, and the standard first unpacking
exercise. (Ch. 15)

**Java Debug Wire Protocol (JDWP)**: The protocol a Java virtual machine speaks to a debugger. Enabled with the JVM option -agentlib:jdwp, it underlies attaching to a running Java process; it has no authentication, so an exposed JDWP port is a remote-code-execution risk. (Ch. 15)

**GDB Enhanced Features (GEF)**: A set of GDB commands for reverse engineers, and the debugger add-on
packaged for Kali. (Ch. 15)

**Library to Instrument Executable Formats (LIEF)**: A library for parsing, modifying and rebuilding
PE, ELF and Mach-O files. (Ch. 15)

**Internet Services Simulation Suite (INetSim)**: Simulates common internet services so a sample
believes it reached the network. (Ch. 15)

**Interactive Disassembler (IDA)**: The commercial disassembler from Hex-Rays. This expansion is the
established one in the literature; Hex-Rays' current product pages do not spell the acronym out.
(Ch. 15)

**YARA and capa**: Product names, not acronyms. YARA's documentation and repository publish no
expansion, and capa's repository describes what the tool does without giving one. Several expansions
circulate for YARA online; none is authoritative. Ghidra, radare2, rizin, Cutter and REMnux are
likewise names. (Ch. 15, Ch. 16)

### Standards, law and certification

These entries cover the knowledge base, statute, and certification that the reverse engineering material refers to, noting where an expansion is established in the literature rather than published by its owner and where a name is not an acronym at all.

**Adversarial Tactics, Techniques, and Common Knowledge (ATT&CK)**: MITRE's knowledge base of
adversary behavior. This expansion is established in the literature; MITRE's current public pages
describe the knowledge base without spelling the letters out. MITRE is the organization's name rather
than an acronym. (Ch. 12, Ch. 15)

**Digital Millennium Copyright Act (DMCA)**: The United States statute containing the
anti-circumvention rules codified at 17 U.S.C. 1201, including the interoperability provision at
1201(f). Separate from the Computer Fraud and Abuse Act (CFAA). (Ch. 15, Ch. 18)

**GIAC Reverse Engineering Malware (GREM)**: The certification aligned with SANS FOR610. GIAC is the
Global Information Assurance Certification body; SANS expands, on its own about page, to SysAdmin,
Audit, Network, Security. (Ch. 15)

### Notation that is not an acronym

A leading **0x** marks a hexadecimal value. **Little endian** storage means the least significant
byte comes first, so the bytes 4D 5A 90 00 read as the 32-bit value 0x00905A4D. **Bits per byte** is
the unit of Shannon entropy, running from 0 when every byte in a block is identical to 8 when all 256
values occur equally often. On Windows API names an **A** suffix takes single-byte strings and a **W**
suffix takes UTF-16, which tells an analyst which string encoding to search for. In assembly,
**square brackets** denote the memory at an address rather than the address itself. (Ch. 15)


## Cryptography Terms from the FOLDOC Category

The Free On-line Dictionary of Computing (FOLDOC, foldoc.org) groups 45 entries under cryptography. Each is
defined below in current terms, with the section of Chapter 2 that covers it. Several FOLDOC entries were
written in the 1990s and are now outdated or wrong; where that matters, the entry says what changed.

**Advanced Encryption Standard (AES)**: NIST's standard block cipher, FIPS 197 (2001, editorially updated
2023), a 128-bit block with 128-, 192-, or 256-bit keys. (Sec. 2.5, 2.6)

**Brute force attack**: Trying every possible key or password until one works. Key length sets the cost;
quantum search (Grover's algorithm) gives only a quadratic speedup, so NIST still treats AES-128 as secure.
(Sec. 2.5, 2.17)

**CA**: See Certificate Authority.

**CAPI**: Microsoft CryptoAPI, the Windows cryptography interface introduced in 1996 with Windows NT 4.0
and Internet Explorer 3.0. Microsoft has deprecated it in favor of CNG (Cryptography API: Next Generation).
FOLDOC's entry is only a stub. (Sec. 2.18)

**Certificate Authority (CA)**: A trusted party that signs certificates binding names to public keys. Public
TLS certificates may be valid for at most 200 days since March 15, 2026, falling to 47 days by 2029.
(Sec. 2.13)

**Ciphertext**: The encrypted form of a message. (Sec. 2.1)

**Clipper**: The 1993 US key-escrow chip that encrypted phone calls with Skipjack and sent a Law
Enforcement Access Field; Matt Blaze showed in 1994 that the field could be forged, and the program was
abandoned. (Sec. 2.16)

**Code**: A system that replaces words or phrases of varying length with arbitrary symbol groups from a
codebook, as opposed to a cipher, which works on fixed units such as letters or bits. (Sec. 2.1)

**Codes**: See Code. In computing, code also means program text or an encoding; the cryptographic sense is
the codebook sense. (Sec. 2.1)

**Cryppie**: Hacker slang (from the Jargon File) for a cryptographer or anyone who builds cryptographic
software or hardware. (Sec. 2.1)

**Cryptography**: The practice of protecting information so that only authorized parties can read or alter
it undetected: confidentiality, integrity, authentication, and non-repudiation. (Sec. 2.1)

**Cryptology**: The science covering both cryptography and cryptanalysis. (Sec. 2.1)

**Decode**: To convert data back from an encoding such as Base64. Decoding uses no key and provides no
secrecy; it is not decryption. (Sec. 2.1)

**Decryption**: Recovering plaintext from ciphertext with the right key. (Sec. 2.1)

**Diffie-Hellman**: The 1976 key-agreement method by Whitfield Diffie and Martin Hellman that lets two parties
derive a shared secret over a public channel. It is key agreement, not encryption, although FOLDOC calls it
an encryption key exchange algorithm. (Sec. 2.11)

**Digital envelope**: Hybrid encryption: the message is encrypted with a fresh symmetric key, and that key is
encrypted with each recipient's public key, as in PKCS #7 and CMS. FOLDOC's entry has no definition.
(Sec. 2.10)

**Digital signature**: A value computed with a private key that anyone can verify with the matching public
key, proving origin and integrity. (Sec. 2.13)

**Digital Signature Standard (DSS)**: NIST FIPS 186. The current FIPS 186-5 (2023) approves RSA, ECDSA, and
EdDSA and no longer approves DSA for new signatures; FOLDOC still describes the original DSA-only standard.
(Sec. 2.13)

**Encode**: To change data into another representation, such as Base64 or UTF-8, with no key and no secrecy.
(Sec. 2.1)

**Encryption**: Transforming plaintext into ciphertext with a key so that only key holders can read it.
(Sec. 2.1)

**Enigma**: The German rotor cipher machine of the Second World War, first broken by Polish cryptanalysts in
1932 and attacked at Bletchley Park with the Bombe. (Sec. 2.2)

**International Data Encryption Algorithm (IDEA)**: A 64-bit block cipher with a 128-bit key by Xuejia Lai and
James Massey (1991 to 1992), used in PGP 2.x. Its 64-bit block makes it a legacy cipher today, despite
FOLDOC's claim that it is the most secure block cipher. (Sec. 2.5)

**Key**: The secret (or, for public keys, published) parameter that controls a cryptographic operation.
Security must rest on the key alone. (Sec. 2.1, 2.16)

**Keyed-Hashing Message Authentication (HMAC)**: The message authentication code of RFC 2104 (1997), titled
HMAC: Keyed-Hashing for Message Authentication, built from a hash function and a shared key. (Sec. 2.8)

**One-way function**: A function that is easy to compute and hard to invert. Hash functions are designed to
be one-way; no one has proved that one-way functions exist. (Sec. 2.7, 2.10)

**Plaintext**: The readable message before encryption. (Sec. 2.1)

**Pretty Good Privacy (PGP)**: Phil Zimmermann's 1991 e-mail encryption program, standardized as OpenPGP
(RFC 9580, 2024). It uses a web of trust instead of central certificate authorities. (Sec. 2.13)

**Private key**: The secret half of an asymmetric key pair, held by its owner, who uses it to decrypt or
sign. FOLDOC also uses the term for a symmetric key and says it is known only to the sender; both uses
conflict with NIST and IETF usage. (Sec. 2.16)

**Private-key cryptography**: An older name for symmetric-key cryptography; prefer the latter. (Sec. 2.5,
2.16)

**PSK**: Pre-shared key, a symmetric key established before a protocol runs, used in WPA2-Personal, TLS 1.3
resumption, and IPsec. FOLDOC's entry is only a stub. (Sec. 2.16)

**Public-Key Cryptography Standards (PKCS)**: The RSA Laboratories standards series (PKCS #1 to #15), now
maintained mostly as IETF RFCs, with PKCS #11 at OASIS. (Sec. 2.10)

**Public-key encryption**: Encryption with a recipient's public key so that only the matching private key can
decrypt; in practice it wraps a symmetric key inside a digital envelope. (Sec. 2.10)

**Public Key Infrastructure (PKI)**: The certificate authorities, certificates, revocation, and policies that
bind public keys to identities. (Sec. 2.13)

**RC4**: Ron Rivest's 1987 stream cipher, leaked in 1994, broken in WEP, and prohibited in TLS by RFC 7465
(2015). (Sec. 2.5)

**RSA**: The public-key algorithm of Rivest, Shamir, and Adleman (1977), based on the difficulty of factoring.
Its US patent expired on September 20, 2000. (Sec. 2.10)

**RSA Data Security, Inc.**: The company the RSA inventors founded in 1982 to license the patent. It became
RSA Security, then part of EMC and Dell, and has been privately held since 2020. (Sec. 2.10)

**RSA encryption**: Encryption with the RSA function; secure use requires OAEP padding (PKCS #1, RFC 8017),
never textbook RSA. (Sec. 2.10)

**SkipJack**: The NSA's 64-bit block cipher with an 80-bit key, used in the Clipper chip and declassified in
1998. NIST has disallowed it for encryption since 2010. (Sec. 2.5, 2.16)

**Symmetric key cryptography**: Cryptography in which the same key encrypts and decrypts, such as AES or
ChaCha20. (Sec. 2.5)

**Synchronous key encryption**: Not a standard term. FOLDOC's 1994 entry uses it for what is actually
public-key encryption; use symmetric or asymmetric instead. (Sec. 2.16)

**SYSKEY**: The Windows system key, added in Windows NT 4.0 Service Pack 3 (1997), that encrypts password
hashes in the SAM. Microsoft removed the syskey.exe utility starting with Windows 10 version 1709, but the
key itself still protects the hashes. (Sec. 2.9)

**System Account Manager**: FOLDOC's name for the Windows Security Account Manager (SAM), the registry
database that stores local account password hashes. Microsoft calls it the Security Account Manager.
(Sec. 2.9)

**Trap-door function**: A one-way function that is easy to invert with a secret, the trapdoor, such as the
factors of an RSA modulus. (Sec. 2.10)

**Triple DES**: DES applied three times. NIST disallowed it for encryption after December 31, 2023 because of
its 64-bit block (the Sweet32 attack) and speed. (Sec. 2.5)

**Tunny Emulator**: FOLDOC's entry for the British Tunny machine, a relay-based copy of the German Lorenz
cipher built at Dollis Hill that printed plaintext once Colossus had found the wheel settings. A working
reconstruction has been on display at The National Museum of Computing since 2011. (Sec. 2.2)
