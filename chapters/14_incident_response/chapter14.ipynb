## 14.1 The Lifecycle

NIST SP 800-61 describes incident handling as a cycle: preparation; detection and analysis; containment,
eradication, and recovery; and post-incident activity {cite}`nist_sp800_61`. The cycle emphasizes that
response is not a single event but an ongoing capability, and that what you do before an incident
largely determines how well you handle it.

## 14.2 Preparation

Preparation is the phase with the highest return. It includes building the team, defining roles and
escalation paths, writing playbooks, ensuring logging is adequate, and rehearsing through tabletop
exercises. An organization that prepares responds calmly and quickly; one that does not improvises under
pressure, usually poorly.

## 14.3 Detection, Containment, Eradication, Recovery

Detection and analysis confirm that an incident is real and scope its impact, often using indicators of
compromise. Containment limits damage, sometimes choosing short-term isolation before a fuller fix.
Eradication removes the adversary's access and artifacts. Recovery restores systems to normal operation
while monitoring for recurrence. Moving too fast to recover before eradication is complete is a common
and costly mistake.

## 14.4 Communication and Lessons Learned

Incidents are as much about communication as technology. Clear internal coordination, appropriate
external and regulatory notification, and disciplined messaging all matter. After resolution, a
blameless post-incident review captures what happened and what to improve, feeding back into
preparation. Treating mistakes as system problems rather than individual failures produces honest,
useful reviews.

## 14.5 Why This Matters

Detection and prevention will eventually be bypassed, so the maturity of an organization's response
often decides whether an incident becomes a minor event or a headline. Response capability is also
increasingly a regulatory and contractual expectation.

## 14.6 News in Focus

Regulatory regimes increasingly impose breach-notification deadlines, sometimes measured in days or even
hours for certain sectors. These requirements have pushed organizations to formalize incident response
and to rehearse it, since meeting tight notification windows is impossible without preparation.

## 14.7 Worked Example: Response Metrics

The cell computes mean time to detect and mean time to respond across several incidents, the metrics
teams track to measure and improve their capability.


````python
incidents = [
    # (detect_minutes_after_start, respond_minutes_after_detect)
    (120, 45),
    (30, 20),
    (480, 90),
    (15, 10),
    (240, 60),
]

mttd = sum(d for d, _ in incidents) / len(incidents)
mttr = sum(r for _, r in incidents) / len(incidents)

print(f"Incidents analyzed: {len(incidents)}")
print(f"Mean Time to Detect (MTTD):  {mttd:.0f} minutes")
print(f"Mean Time to Respond (MTTR): {mttr:.0f} minutes")
print(f"Average total dwell before response begins: {mttd + mttr:.0f} minutes")

worst = max(incidents, key=lambda x: x[0])
print(f"\nLongest detection time: {worst[0]} minutes -> prioritize improving detection coverage.")
````
