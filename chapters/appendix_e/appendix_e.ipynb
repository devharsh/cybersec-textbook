# Appendix E: Selected Works by the Author (Mapped to Chapters)

Several chapters of this book draw on, or connect to, the author's own research and writing. This
appendix lists those works and maps each to the chapter(s) it most relates to, so that readers who want to
go deeper on a topic can follow it to a primary source. Entries are grouped by theme; the chapter tag in
parentheses shows where the topic appears in this book. Bibliographic details follow the author's public
publication record; readers should consult the cited venue or DOI for the authoritative version.
