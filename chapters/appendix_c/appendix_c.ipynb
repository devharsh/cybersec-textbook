# Appendix C: Certification Mapping

This appendix demonstrates how the twenty chapters of this textbook cover the official
examination domains of eight widely held security certifications: **(ISC)2 CISSP**,
**CompTIA Security+ (SY0-701)**, **EC-Council CEH v13**, **ISACA CISA**,
**(ISC)2 CGRC** (formerly CAP, aligned to the NIST Risk Management Framework),
**(ISC)2 SSCP**, **(ISC)2 Certified in Cybersecurity (CC)**, and
**GIAC GREM** (Reverse Engineering Malware, aligned to SANS FOR610).

Domain titles and weightings below reflect the published exam outlines current as of
2026. Always confirm against the certifying body latest outline before sitting an
exam, because vendors revise objectives periodically; for example CompTIA CySA+ moved
from CS0-003 to CS0-004 in mid 2026, and the (ISC)2 CC outline effective September 2026
folds AI security into every domain. The mapping helps an instructor or self-study
learner locate the chapters that develop each domain in depth.

**Course alignment.** The authors four Fall 2026 courses each map to one of these
credentials: LaGuardia MAC 237 Computer Security to the (ISC)2 CC (Section C.7),
MAC 246 Advanced Network Security to CompTIA Security+ SY0-701 (Section C.2),
MAC 247 Advanced Systems Security to the (ISC)2 SSCP (Section C.6), and Bowie State
COSC 332 Software Reverse Engineering to the GIAC GREM (Section C.8).

A chapter listed in **bold** is the primary home of that domain; lighter listings
indicate supporting coverage.
