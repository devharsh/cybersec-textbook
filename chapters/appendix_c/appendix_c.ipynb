# Appendix C: Certification Mapping

This appendix suggests which chapters support preparation for common certifications. It is a study aid,
not an official syllabus; always consult the current exam objectives published by each certifying body,
since objectives change over time.

## CompTIA Security+

A broad, entry-level certification covering foundational concepts.

| Topic Area | Chapters |
|---|---|
| General security concepts | 1, 5 |
| Cryptography | 2 |
| Network security | 3, 11 |
| Threats and social engineering | 4 |
| Detection and monitoring | 12 |
| Incident response and forensics | 13, 14 |
| Governance and risk | 5 |

## Certified Ethical Hacker (CEH)

Focused on offensive techniques within an authorized framework.

| Topic Area | Chapters |
|---|---|
| Methodology and ethics | 6 |
| Reconnaissance | 7 |
| Scanning and enumeration | 8 |
| Exploitation concepts | 9 |
| Web application attacks | 10 |
| Malware | 15 |
| Practical skill building | 16 |

## CISSP

A management-oriented certification spanning eight domains.

| CISSP Domain | Chapters |
|---|---|
| Security and Risk Management | 1, 5 |
| Asset Security | 5, 13 |
| Security Architecture and Engineering | 2, 11 |
| Communication and Network Security | 3, 11 |
| Identity and Access Management | 10, 11 |
| Security Assessment and Testing | 6, 7, 8 |
| Security Operations | 12, 13, 14, 15 |
| Software Development Security | 9, 10 |

Use these mappings to assemble a focused reading plan, then practice with the labs and the
capture-the-flag material in Chapter 16 to convert knowledge into skill.
