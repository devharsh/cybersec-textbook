# Appendix C: Certification Mapping

This appendix demonstrates how the twenty chapters of this textbook cover the official
examination domains of eight widely held security certifications: **(ISC)2 CISSP**,
**CompTIA Security+ (SY0-701)**, **EC-Council Certified Ethical Hacker (CEH) v13**, **ISACA CISA**, **(ISC)2 Certified in Governance, Risk and Compliance (CGRC)** (formerly CAP, aligned to the NIST Risk Management Framework), **(ISC)2 Systems Security Certified Practitioner (SSCP)**, **(ISC)2 Certified in Cybersecurity (CC)**, and
**GIAC GREM** (Reverse Engineering Malware, aligned to SANS FOR610). It also maps the pilot exam outline of a new credential, the **ISC2 AI security
certification**, which ISC2 plans to launch in 2027 (Section C.9).

Domain titles and weightings below reflect the published exam outlines current as of
2026. Always confirm against the certifying body latest outline before sitting an
exam, because vendors revise objectives periodically; for example the CompTIA Cybersecurity Analyst credential, CySA+, moved
from CS0-003 to CS0-004 in mid 2026, and the (ISC)2 CC outline effective September 1,
2026 renamed and reweighted all five domains and folds AI security into every one of
them. The mapping helps an instructor or self-study
learner locate the chapters that develop each domain in depth.

**Course alignment.** The authors four Fall 2026 courses each map to one of these
credentials: LaGuardia MAC 237 Computer Security to the (ISC)2 CC (Section C.7),
MAC 246 Advanced Network Security to CompTIA Security+ SY0-701 (Section C.2),
MAC 247 Advanced Systems Security to the (ISC)2 SSCP (Section C.6), and Bowie State
COSC 332 Software Reverse Engineering to the GIAC GREM (Section C.8).

A chapter listed in **bold** is the primary home of that domain; lighter listings
indicate supporting coverage.

## C.9 ISC2 AI Security Certification (Pilot) - 6 Domains

ISC2 announced on July 15, 2026 that it was developing a certification in AI security, and it released the pilot exam outline on October 26, 2026, during its Security Congress in Denver. The credential does not have an official name yet; ISC2 calls it its AI security certification. More than 3,000 selected volunteers will sit a pilot exam at Pearson Professional Centers from mid-December 2026 through January 24, 2027. ISC2 plans to set the passing standard in 2027 and to launch the certification that year if the pilot succeeds. Treat every detail below as provisional until then.

The proposed experience requirement is four years of cumulative, paid, full-time work in cybersecurity or AI-related fields. Two of those years must be in cybersecurity, and one must be in two or more of the outline's domains. A bachelor's or master's degree in computer science, information technology, or another science, technology, engineering, or mathematics (STEM) field can waive one year. An active CISSP, or one of the ISC2 credentials Information Systems Security Architecture Professional (ISSAP), Information Systems Security Engineering Professional (ISSEP), Information Systems Security Management Professional (ISSMP), Certified Cloud Security Professional (CCSP), or Certified Secure Software Lifecycle Professional (CSSLP), replaces the entire experience requirement.
