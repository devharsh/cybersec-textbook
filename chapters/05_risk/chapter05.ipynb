## 5.1 What Risk Management Is

Security resources are always finite, so risk management is the discipline of deciding where to spend
them. NIST SP 800-30 frames assessment as identifying threats and vulnerabilities, determining
likelihood and impact, and producing a prioritized view of risk {cite}`nist_sp800_30`. The output is
not a guarantee of safety but a defensible basis for decisions.

## 5.2 Qualitative and Quantitative Assessment

Qualitative assessment ranks risks on ordinal scales such as low, medium, and high, often using a risk
matrix that crosses likelihood with impact. It is fast and communicates well to leadership.
Quantitative assessment assigns monetary values and probabilities, enabling calculations such as ALE
and ROSI. Most mature programs blend both: quantitative where data exists and qualitative where it does
not.

## 5.3 Risk Treatment

Once a risk is understood there are four treatment options. **Mitigate** reduces likelihood or impact
through controls. **Transfer** shifts financial consequences, for example through insurance or
contracts. **Avoid** eliminates the activity that creates the risk. **Accept** acknowledges the risk
when the cost of treatment exceeds the expected loss. Every accepted risk should be documented and owned
by someone with the authority to accept it.

## 5.4 Governance and Compliance

Governance sets direction through policies, roles, and accountability. Standards such as ISO/IEC 27001
describe a management system for continual improvement, while regulations impose sector-specific duties.
Compliance is necessary but not sufficient: meeting a checklist does not guarantee security, yet a sound
program will usually satisfy most compliance requirements as a byproduct.

## 5.5 Why This Matters

Technical staff who can speak the language of risk and translate a vulnerability into expected business
impact are far more effective at securing budget and driving change than those who present only
technical severity. Risk thinking is the bridge between the lab and the boardroom.

## 5.7 Worked Numerical Example: Return on Security Investment

ROSI compares the loss a control is expected to prevent against the cost of the control:

$$ROSI = \frac{(ALE_{before} - ALE_{after}) - CostOfControl}{CostOfControl}$$


````python
ale_before = 120_000      # expected annual loss without the control
mitigation = 0.75         # control removes 75% of expected loss
cost_of_control = 30_000  # annualized cost of the control

ale_after = ale_before * (1 - mitigation)
benefit = ale_before - ale_after
rosi = (benefit - cost_of_control) / cost_of_control * 100

print(f"ALE before control: ${ale_before:,.2f}")
print(f"ALE after control:  ${ale_after:,.2f}")
print(f"Annual benefit:     ${benefit:,.2f}")
print(f"Cost of control:    ${cost_of_control:,.2f}")
print(f"ROSI:               {rosi:.1f}%")
print("Investment is justified" if rosi > 0 else "Investment is not justified")
````

## 5.7b Worked Numerical Example: Risk Reduction from a Control

A second common calculation expresses the value of a control as the reduction in expected loss it produces.
Suppose a data store faces an Annualized Loss Expectancy of $2,000,000 per breach. Historical data suggests
the probability of a breach is 0.40 while the data is left unencrypted and 0.08 once encryption is applied.
The expected annual loss in each case is the probability multiplied by the loss, and the risk reduction is
the difference. This isolates the benefit term that also feeds the ROSI formula above.

````python
ale_per_breach = 2_000_000
p_unencrypted = 0.40
p_encrypted = 0.08

expected_unencrypted = p_unencrypted * ale_per_breach
expected_encrypted = p_encrypted * ale_per_breach
risk_reduction = expected_unencrypted - expected_encrypted

print(f"Expected annual loss, unencrypted: ${expected_unencrypted:,.0f}")
print(f"Expected annual loss, encrypted:   ${expected_encrypted:,.0f}")
print(f"Risk reduction (annual benefit):   ${risk_reduction:,.0f}")
````
