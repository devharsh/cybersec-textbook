## Key Terms

- **Phishing**: Fraudulent messages designed to elicit credentials or actions.
- **Pretexting**: Inventing a scenario to obtain information.
- **BEC**: Business Email Compromise.
- **MFA**: Multi-Factor Authentication.
- **Vishing**: Voice phishing over the telephone.

## 4.1 Why People Are the Target

Technical controls have matured to the point that attacking software directly is often harder than
persuading a person to act against their own interest. Social engineering exploits universal cognitive
tendencies rather than software bugs. The Verizon report repeatedly finds that the human element,
including error and social attacks, is involved in a large fraction of breaches {cite}`verizon_dbir_2024`.
This is why security awareness is treated as a control in its own right.

## 4.2 Principles of Influence

Attackers lean on well-documented principles: authority (appearing to be a manager or official),
urgency (forcing fast decisions that bypass scrutiny), scarcity (a limited-time offer), social proof
(everyone else has complied), and reciprocity (a small favor that invites a return). Recognizing these
levers in an unexpected message is the single most useful defensive habit a person can develop.

## 4.3 Attack Techniques

**Phishing** casts a wide net with generic lures. **Spear phishing** targets a specific individual with
tailored detail, and **whaling** targets executives. **Business email compromise** impersonates a
trusted party to redirect payments and is among the costliest categories of fraud. **Pretexting**
builds a believable backstory, **baiting** leaves tempting media or offers, and **tailgating** follows
an authorized person through a physical door. **Vishing** and SMS-based smishing extend these tactics
to other channels.

## 4.4 Defenses

Defense is layered. Technical controls include email authentication such as SPF, DKIM, and DMARC,
attachment sandboxing, and phishing-resistant multi-factor authentication. Process controls include
out-of-band verification for payment changes. Human controls include realistic, frequent training and a
blame-free reporting culture so that employees report suspected attacks quickly rather than hiding
mistakes.

## 4.5 Why This Matters

You can deploy the strongest cryptography in the world and still lose if an employee hands over a
password. Treating people as part of the system, with their own failure modes and their own defenses,
is essential to a complete security program.

## 4.6 News in Focus

A widely reported 2020 incident at a major social media platform involved attackers using social
engineering by telephone to manipulate employees into providing access to internal tools, which were
then used to hijack high-profile accounts. The episode showed that even sophisticated technical
organizations remain exposed through their people and internal processes.

## 4.7 Worked Numerical Example: Campaign Click Rate

A simulated phishing exercise sends messages to staff. The code computes click-through and credential
submission rates and a simple risk-reduction estimate after training.


````python
sent = 1200
clicked = 168
submitted_credentials = 54

click_rate = clicked / sent * 100
submit_rate = submitted_credentials / sent * 100
submit_given_click = submitted_credentials / clicked * 100

print(f"Messages sent:           {sent}")
print(f"Click-through rate:      {click_rate:.1f}%")
print(f"Credential submit rate:  {submit_rate:.1f}%")
print(f"Submit given a click:    {submit_given_click:.1f}%")

# After training, suppose click rate drops to 6%. Estimate exposed users avoided.
post_training_rate = 0.06
avoided = clicked - round(sent * post_training_rate)
print(f"\nEstimated users no longer clicking after training: {avoided}")
````
